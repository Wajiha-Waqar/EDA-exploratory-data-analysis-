```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
iris = pd.read_csv("1) iris.csv")
```

```python
# Display first 5 rows
print(iris.head())
```

```python
# Display last 5 rows
print(iris.tail())
```

```python
# Dataset shape
print("Shape:", iris.shape)
```

```python
# Column names
print("Columns:", iris.columns.tolist())
```

```python
# Data types
print("Data Types:\n", iris.dtypes)
```

```python
# Check for missing values
print("Missing Values:\n", iris.isnull().sum())
```

```python
# General statistics
print(iris.describe())
```

```python
# Group by species
print(iris.groupby("species").describe())
```

```python
iris.hist(figsize=(8, 8))
plt.suptitle("Feature Distributions", y=1.02)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=iris, x="species", y="sepal_length")
plt.title("Sepal Length by Species")
plt.show()
```

```python
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species")
plt.title("Petal Length vs Petal Width")
plt.show()
```

```python
sns.pairplot(iris, hue="species")
plt.suptitle("Pair Plot of Iris Features", y=1.02)
plt.show()
```

```python
# Compute correlation matrix
corr = iris.corr(numeric_only=True)

# Plot heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Using boxplots
sns.boxplot(data=iris[["sepal_width"]])
plt.title("Outliers in Sepal Width")
plt.show()
```
